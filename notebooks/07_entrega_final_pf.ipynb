### 2.1 Distribución de la variable objetivo

###  Interpretación de la Relación con Variables Externas

 **Matriz de Correlación:**

Se evidencian tres patrones principales:

- Existe una **alta correlación entre la variable objetivo y su rezago (`tasa_lag1`)**, lo que confirma una fuerte persistencia temporal del delito.
- Se observa **correlación moderada con la densidad poblacional (`log_densidad`)**, indicando la influencia de factores estructurales territoriales.
- Las variables económicas presentan **correlaciones débiles con el delito pero muy altas entre sí**, lo que evidencia multicolinealidad y refuerza que el fenómeno no depende de un único factor económico.

Los resultados sugieren un fenómeno dominado por dinámica temporal y estructura territorial, con influencia económica secundaria.

 **Delito vs Inflación:**

El gráfico de dispersión de "Delito vs Inflación" (Indice_IPC) presenta una gran dispersión de puntos. Aunque la línea de tendencia puede indicar una ligera pendiente positiva, la variabilidad de los datos alrededor de esta línea es muy alta. Esta relación no es consistente ni dominante y no permite predecir el delito basándose solo en la inflación.

  **Delito vs Costo de Vida:**

Aquí se observa una asociación positiva más discernible en comparación con otras variables. Esto podría interpretarse como una presión económica creciente que potencialmente impulsa el aumento de ciertos tipos de delitos, aunque la causalidad no se establece solo con esta observación.

 **Delito vs Salarios:**

Se observa aquí una relación muy débil. Esto indica que los cambios en la variación anual de los salarios no tienen una influencia clara o directa sobre la tasa de delitos contra la propiedad en el dataset analizado.

 **Delito vs Empleo en Construcción:**

Similar a la relación con los salarios, el gráfico de "Delito vs Empleo construcción" también exhibe una correlación muy débil o nula.

Los hallazgos descriptivos refuerzan la idea de que los delitos contra la propiedad son un fenómeno **multicausal y complejo**.


 **Delito vs tasa rezagada (`tasa_lag1`)**

El gráfico muestra una relación claramente positiva y aproximadamente lineal entre la tasa de delitos actual y su valor en el período anterior.
- Existe una fuerte alineación de los puntos en torno a una tendencia creciente
- La dispersión es relativamente acotada en valores medios
- A medida que aumentan los valores, se observa mayor variabilidad (presencia de outliers)
- La tasa de delitos presenta una **fuerte persistencia temporal**.

Es decir, los niveles actuales de delito están altamente condicionados por los niveles del período anterior, lo que sugiere que el fenómeno posee una dinámica propia difícil de revertir en el corto plazo.

 **Delito vs densidad poblacional (`log_densidad`)**

Se observa una relación notoria entre la tasa de delitos y la densidad poblacional transformada; con una tendencia positiva general, aunque menos marcada que en el caso del lag.
- Existe una alta dispersión para niveles bajos y medios de densidad
- En valores altos de densidad aparecen concentraciones con mayores niveles de delito
- Se detectan algunos outliers en zonas de alta densidad
- A mayor densidad poblacional, mayor tiende a ser la tasa de delitos contra la propiedad.

Este resultado es coherente con la idea de que la densidad actúa como un **factor estructural que genera condiciones propicias**, pero no suficiente para explicar completamente el fenómeno.

```python
# ============================================================
# MAPA INTERACTIVO EN FOLIUM
# ------------------------------------------------------------
# Esta versión incluye:
# - filtros interactivos por año y provincia
# - tamaño de círculos normalizado según la tasa
# - color de círculos según intensidad de la tasa
# - tooltip + popup detallado
# - título incrustado sobre el mapa
# - control de capas
# - cluster de marcadores para zonas densas
# - leyenda de color
# - leyenda visual de tamaño


# ============================================================
# 1. WIDGETS DE CONTROL
# ------------------------------------------------------------
# Se crean dos controles:
# - un slider para elegir el año
# - un dropdown para elegir una provincia o ver todo el país
# ============================================================

anio_widget = widgets.SelectionSlider(
    options=sorted(df_mapa_sup_final["anio"].dropna().unique()),
    value=2024,
    description="Año",
    continuous_update=False,
    layout=widgets.Layout(width="320px")
)

provincia_widget = widgets.Dropdown(
    options=["Todas"] + sorted(df_mapa_sup_final["provincia_nombre"].dropna().unique()),
    value="Todas",
    description="Provincia",
    layout=widgets.Layout(width="320px")
)

# Output donde se dibujará el mapa cada vez que cambie un filtro
output = widgets.Output()


# ============================================================
# 2. FUNCIONES AUXILIARES
# ============================================================

def calcular_radio(valor, vmin, vmax, rmin=4, rmax=16):
    """
    Escala el tamaño del círculo entre rmin y rmax
    usando normalización lineal sobre la variable elegida.
    """
    if vmax == vmin:
        return (rmin + rmax) / 2
    return rmin + ((valor - vmin) / (vmax - vmin)) * (rmax - rmin)


def formatear_numero(valor, decimales=2):
    """
    Formatea números para el popup de forma prolija.
    """
    return f"{valor:,.{decimales}f}".replace(",", "X").replace(".", ",").replace("X", ".")


def agregar_titulo(mapa, anio, provincia):
    """
    Agrega un título HTML incrustado sobre el mapa.
    """
    subtitulo = "Total país" if provincia == "Todas" else provincia

    titulo_html = f"""
    <div style="
        position: fixed;
        top: 10px;
        left: 50px;
        z-index: 9999;
        background-color: rgba(17,17,17,0.88);
        color: white;
        padding: 10px 14px;
        border-radius: 8px;
        font-size: 16px;
        font-weight: bold;
        box-shadow: 0 0 6px rgba(0,0,0,0.45);
        border: 1px solid rgba(255,255,255,0.08);
    ">
        Tasa de delitos contra la propiedad por departamento - Año {anio}<br>
        <span style="font-size: 13px; font-weight: normal; color: #d0d0d0;">
            Cobertura: {subtitulo}
        </span>
    </div>
    """
    mapa.get_root().html.add_child(folium.Element(titulo_html))


def agregar_leyenda_tamanio(mapa):
    """
    Agrega una leyenda visual para interpretar el tamaño de los círculos.
    Es una referencia conceptual de 'menor' a 'mayor' tasa.
    """
    leyenda_html = """
    <div style="
        position: fixed;
        bottom: 30px;
        left: 30px;
        z-index: 9999;
        background-color: rgba(17,17,17,0.92);
        color: white;
        padding: 12px 14px;
        border-radius: 8px;
        font-size: 12px;
        box-shadow: 0 0 6px rgba(0,0,0,0.45);
        border: 1px solid rgba(255,255,255,0.08);
        width: 185px;
    ">
        <div style="font-weight: bold; margin-bottom: 8px;">
            Tamaño del marcador
        </div>

        <div style="display: flex; align-items: center; margin-bottom: 6px;">
            <div style="
                width: 8px; height: 8px; border-radius: 50%;
                background: #ffb74d; margin-right: 10px;
                border: 1px solid #111;
            "></div>
            <span>Menor tasa relativa</span>
        </div>

        <div style="display: flex; align-items: center; margin-bottom: 6px;">
            <div style="
                width: 14px; height: 14px; border-radius: 50%;
                background: #f4511e; margin-right: 10px;
                border: 1px solid #111;
            "></div>
            <span>Tasa intermedia</span>
        </div>

        <div style="display: flex; align-items: center;">
            <div style="
                width: 22px; height: 22px; border-radius: 50%;
                background: #8e0038; margin-right: 10px;
                border: 1px solid #111;
            "></div>
            <span>Mayor tasa relativa</span>
        </div>
    </div>
    """
    mapa.get_root().html.add_child(folium.Element(leyenda_html))


# ============================================================
# 3. FUNCIÓN PRINCIPAL DE CREACIÓN DEL MAPA
# ============================================================

def crear_mapa_folium_completo(df_base, anio, provincia="Todas", usar_cluster=True):

    # --------------------------------------------------------
    # A. FILTRADO DEL DATASET
    # --------------------------------------------------------
    df_filtered = df_base[df_base["anio"] == anio].copy()

    if provincia != "Todas":
        df_filtered = df_filtered[df_filtered["provincia_nombre"] == provincia].copy()

    # Filtrado defensivo para asegurar que las coordenadas y
    # la variable visual principal estén disponibles
    df_filtered = df_filtered.dropna(
        subset=["lat", "lon", "tasa_delitos_propiedad_100k_v2"]
    )

    if df_filtered.empty:
        return None

    # --------------------------------------------------------
    # B. CENTRO Y ZOOM DEL MAPA
    # --------------------------------------------------------
    # El centro se calcula dinámicamente sobre el subconjunto filtrado
    center_lat = df_filtered["lat"].mean()
    center_lon = df_filtered["lon"].mean()

    # Ajuste de zoom según el nivel geográfico observado
    zoom_base = 4 if provincia == "Todas" else 6

    # --------------------------------------------------------
    # C. CREACIÓN DEL MAPA BASE
    # --------------------------------------------------------
    mapa = folium.Map(
        location=[center_lat, center_lon],
        zoom_start=zoom_base,
        tiles="CartoDB dark_matter",
        control_scale=True
    )

    # --------------------------------------------------------
    # D. ESCALA DE COLOR
    # --------------------------------------------------------
    tasa_min = df_filtered["tasa_delitos_propiedad_100k_v2"].min()
    tasa_max = df_filtered["tasa_delitos_propiedad_100k_v2"].max()

    colormap = cm.LinearColormap(
        colors=["#ffe082", "#ffb74d", "#ff7043", "#f4511e", "#d32f2f", "#8e0038"],
        vmin=tasa_min,
        vmax=tasa_max
    )
    colormap.caption = "Tasa de delitos contra la propiedad por 100.000 hab."
    colormap.add_to(mapa)

    # --------------------------------------------------------
    # E. CAPAS DEL MAPA
    # --------------------------------------------------------
    # Capa con los puntos individuales
    capa_puntos = folium.FeatureGroup(name="Departamentos", show=True)

    # Capa cluster opcional para zonas con mucha densidad
    if usar_cluster:
        cluster = MarkerCluster(name="Agrupación de marcadores").add_to(capa_puntos)

    # --------------------------------------------------------
    # F. RECORRIDO DE FILAS Y CREACIÓN DE MARCADORES
    # --------------------------------------------------------
    for _, row in df_filtered.iterrows():
        tasa = row["tasa_delitos_propiedad_100k_v2"]
        radio = calcular_radio(tasa, tasa_min, tasa_max, rmin=4, rmax=16)

        # Popup enriquecido
        popup_html = f"""
        <div style="font-family: Arial; font-size: 13px; width: 270px;">
            <h4 style="margin-bottom: 8px; color: #111;">
                {row['departamento_nombre']}
            </h4>

            <b>Provincia:</b> {row['provincia_nombre']}<br>
            <b>Año:</b> {row['anio']}<br>
            <b>Hechos registrados:</b> {int(row['delitos_propiedad_hechos']) if pd.notna(row['delitos_propiedad_hechos']) else 'N/D'}<br>
            <b>Tasa por 100.000 hab.:</b> {formatear_numero(row['tasa_delitos_propiedad_100k_v2'], 2)}<br>
            <b>Tasa porcentual:</b> {formatear_numero(row['tasa_delitos_propiedad_pct'], 3)}
        </div>
        """

        tooltip_text = (
            f"{row['departamento_nombre']} | "
            f"{row['provincia_nombre']} | "
            f"Tasa: {formatear_numero(row['tasa_delitos_propiedad_100k_v2'], 2)}"
        )

        marcador = folium.CircleMarker(
            location=[row["lat"], row["lon"]],
            radius=radio,
            popup=folium.Popup(popup_html, max_width=320),
            tooltip=tooltip_text,
            color="#111111",
            weight=0.7,
            fill=True,
            fill_color=colormap(tasa),
            fill_opacity=0.82
        )

        if usar_cluster:
            marcador.add_to(cluster)
        else:
            marcador.add_to(capa_puntos)

    # La capa final se agrega al mapa
    capa_puntos.add_to(mapa)

    # --------------------------------------------------------
    # G. ELEMENTOS DE INTERFAZ
    # --------------------------------------------------------
    folium.LayerControl(collapsed=False).add_to(mapa)
    agregar_titulo(mapa, anio, provincia)
    agregar_leyenda_tamanio(mapa)

    return mapa


# ============================================================
# 4. FUNCIÓN DE ACTUALIZACIÓN DEL OUTPUT
# ------------------------------------------------------------
# Cada vez que cambia el año o la provincia, se recrea el mapa
# y se vuelve a mostrar dentro del widget de salida.
# ============================================================

def actualizar_mapa(change=None):
    with output:
        output.clear_output(wait=True)

        mapa = crear_mapa_folium_completo(
            df_base=df_mapa_sup_final,
            anio=anio_widget.value,
            provincia=provincia_widget.value,
            usar_cluster=False              # Colocar "True" para activar la agrupación por cluster
        )

        if mapa is None:
            print("No hay datos para los filtros seleccionados.")
            return

        display(mapa)


# ============================================================
# 5. VINCULACIÓN DE EVENTOS
# ============================================================

anio_widget.observe(actualizar_mapa, names="value")
provincia_widget.observe(actualizar_mapa, names="value")


# ============================================================
# 6. DISPLAY FINAL EN EL NOTEBOOK
# ============================================================

display(widgets.HBox([anio_widget, provincia_widget]))
display(output)

# Primera renderización
actualizar_mapa()
```

### 4.2 Construcción de la base para modelado

```python
# Selección de variables finales para modelado

variables_modelo = [
    "tasa_delitos_propiedad_100k_v2",
    "log_densidad",
    "variacion_anual_cbt",
    "variacion_anual_salarios_pct",
    "variacion_empleo_const_pct",
    "variacion_acceso_internet_pct",
    "tasa_lag1",
    "anio",
    "departamento_key"
]

# Construcción del dataset de modelado
df_modelado = df_mapa_sup_final[variables_modelo].copy()

# Eliminación de nulos
df_modelado = df_modelado.dropna()

# Orden correcto para panel
df_modelado = df_modelado.sort_values(by=["departamento_key", "anio"])

# Verificación
print("Dimensión del dataset de modelado:", df_modelado.shape)
df_modelado.head()
```

### 4.3 Selección de variables

```python
# Variables finales del modelo
features = [
    "log_densidad",
    "variacion_anual_cbt",
    "variacion_anual_salarios_pct",
    "variacion_empleo_const_pct",
    "variacion_acceso_internet_pct",
    "tasa_lag1"
]

target = "tasa_delitos_propiedad_100k_v2"

X = df_modelado[features]
y = df_modelado[target]

print("Variables seleccionadas:", features)
```

### 4.4 Particion temporal en entrenamiento y prueba

```python
# Split temporal

train = df_modelado[df_modelado["anio"] <= 2022].copy()
test = df_modelado[df_modelado["anio"] > 2022].copy()

# Separación de variables
X_train = train[features]
y_train = train[target]

X_test = test[features]
y_test = test[target]

# Verificación
print("Train:", train.shape)
print("Test:", test.shape)

print("Años train:", sorted(train["anio"].unique()))
print("Años test:", sorted(test["anio"].unique()))
```

### 4.5 Modelo Baseline

```python

# Modelo
modelo_lr = LinearRegression()

# Entrenamiento
modelo_lr.fit(X_train, y_train)

# Predicción
y_pred = modelo_lr.predict(X_test)
```

```python
# métricas

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Modelo Baseline")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
# Coeficientes

coeficientes = pd.DataFrame({
    "variable": features,
    "coeficiente": modelo_lr.coef_
}).sort_values(by="coeficiente", ascending=False)

coeficientes
```

### 4.6 Modelo de Efectos Fijos

```python
import statsmodels.formula.api as smf

# Fórmula
formula_fe = """
tasa_delitos_propiedad_100k_v2 ~
log_densidad +
variacion_anual_cbt +
variacion_anual_salarios_pct +
variacion_empleo_const_pct +
variacion_acceso_internet_pct +
tasa_lag1 +
C(departamento_key)
"""

# Entrenamiento
modelo_fe = smf.ols(formula_fe, data=train).fit()

print("R2 (train):", modelo_fe.rsquared)
print("Adj R2:", modelo_fe.rsquared_adj)
```

```python
# Filtrar test para departamentos vistos en train
dept_train = set(train["departamento_key"].unique())
test_fe = test[test["departamento_key"].isin(dept_train)].copy()

y_test_fe = test_fe[target]
y_pred_fe = modelo_fe.predict(test_fe)

# Métricas
mae_fe = mean_absolute_error(y_test_fe, y_pred_fe)
rmse_fe = np.sqrt(mean_squared_error(y_test_fe, y_pred_fe))
r2_fe = r2_score(y_test_fe, y_pred_fe)

print("Modelo FE por departamento")
print("MAE:", mae_fe)
print("RMSE:", rmse_fe)
print("R2:", r2_fe)
```

### 4.7 Modelo Random *Forest*

```python
# Entrenamiento

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
# Predicción
y_pred_rf = rf.predict(X_test)

#Métricas
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest")
print("MAE:", mae_rf)
print("RMSE:", rmse_rf)
print("R2:", r2_rf)
```

```python
# Importancia de variables
importancia = pd.DataFrame({
    "variable": X_train.columns,
    "importancia": rf.feature_importances_
}).sort_values(by="importancia", ascending=False)

importancia
```

### 4.8 Modelo XGBoost

```python
# ================================
# 4.8 Modelo XGBoost
# ================================

xgb = XGBRegressor(
   n_estimators=75,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.8,
    colsample_bytree=1,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb.fit(X_train, y_train)

y_pred_xgb = xgb.predict(X_test)

mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = r2_score(y_test, y_pred_xgb)

print("XGBoost")
print("MAE:", mae_xgb)
print("RMSE:", rmse_xgb)
print("R2:", r2_xgb)
```

```python
# Importancia de variables - XGBoost

importancia_xgb = pd.DataFrame({
    "variable": X_train.columns,
    "importancia": xgb.feature_importances_
}).sort_values(by="importancia", ascending=False)

importancia_xgb
```

El modelo XGBoost presentó el mejor desempeño global del proyecto, obteniendo las menores métricas de error y el mayor coeficiente de determinación. La mejora respecto de Random Forest es moderada, aunque consistente, lo que sugiere que ambos enfoques logran capturar adecuadamente la dinámica temporal y territorial del fenómeno.

La incorporación de boosting secuencial permitió optimizar progresivamente los errores de predicción y mejorar la capacidad de generalización sobre datos no observados.

## 4.9 Explicabilidad del modelo ganador (SHAP)

```python
# ================================
# 4.9 SHAP - Explicabilidad
# ================================

# Muestra para SHAP
X_shap = X_test.sample(
    n=min(500, len(X_test)),
    random_state=42
)

# Explainer para modelo ganador XGBoost
explainer = shap.TreeExplainer(xgb)

# Valores SHAP
shap_values = explainer.shap_values(X_shap)

print("SHAP calculado correctamente.")
```

```python
# ================================
# Importancia global SHAP
# ================================

mean_abs_shap_values = np.mean(
    np.abs(shap_values),
    axis=0
)

shap_importance_df = (
    pd.DataFrame({
        "Variable": X_shap.columns,
        "Impacto_SHAP_medio_absoluto": mean_abs_shap_values
    })
    .sort_values(
        by="Impacto_SHAP_medio_absoluto",
        ascending=True
    )
)

fig_shap_importance = px.bar(
    shap_importance_df,
    x="Impacto_SHAP_medio_absoluto",
    y="Variable",
    orientation="h",
    text="Impacto_SHAP_medio_absoluto",
    title="Importancia global de variables según SHAP",
    template="plotly_dark"
)

fig_shap_importance.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig_shap_importance.update_layout(
    title_x=0.5,
    xaxis_title="Impacto medio absoluto sobre la predicción",
    yaxis_title="Variable",
    margin=dict(l=20, r=40, t=70, b=40)
)

fig_shap_importance.show()
```

```python
# ================================
# SHAP - Dirección del efecto
# ================================

# Convertir SHAP a DataFrame largo
shap_df = pd.DataFrame(
    shap_values,
    columns=X_shap.columns
)

X_shap_reset = X_shap.reset_index(drop=True)

# Pasar a formato largo
shap_long = shap_df.reset_index().melt(
    id_vars="index",
    var_name="Variable",
    value_name="Valor_SHAP"
)

features_long = X_shap_reset.reset_index().melt(
    id_vars="index",
    var_name="Variable",
    value_name="Valor_variable"
)

shap_plot_df = shap_long.merge(
    features_long,
    on=["index", "Variable"]
)

# Ordenar variables según impacto medio absoluto
orden_variables = (
    shap_plot_df
    .assign(abs_shap=lambda x: x["Valor_SHAP"].abs())
    .groupby("Variable")["abs_shap"]
    .mean()
    .sort_values(ascending=True)
    .index
)

shap_plot_df["Variable"] = pd.Categorical(
    shap_plot_df["Variable"],
    categories=orden_variables,
    ordered=True
)

fig_shap_direction = px.scatter(
    shap_plot_df,
    x="Valor_SHAP",
    y="Variable",
    color="Valor_variable",
    color_continuous_scale="Turbo",
    title="Dirección del efecto de las variables según SHAP",
    labels={
        "Valor_SHAP": "Impacto SHAP sobre la predicción",
        "Variable": "Variable",
        "Valor_variable": "Valor de la variable"
    },
    template="plotly_dark",
    opacity=0.65
)

fig_shap_direction.add_vline(
    x=0,
    line_dash="dash",
    line_color="white"
)

fig_shap_direction.update_layout(
    title_x=0.5,
    xaxis_title="Impacto SHAP sobre la predicción",
    yaxis_title="Variable",
    coloraxis_colorbar_title="Valor variable",
    height=650,
    margin=dict(l=20, r=40, t=70, b=40)
)

fig_shap_direction.show()
```

### Interpretación de los resultados SHAP

#### Importancia global de variables

El análisis SHAP confirma que la variable **tasa_lag1** es, por amplio margen, el predictor más relevante dentro del modelo XGBoost. Su impacto medio absoluto supera ampliamente al resto de las variables, lo que indica que la información histórica reciente de la tasa de delitos constituye el principal determinante de las predicciones.

En un segundo nivel de importancia aparece **variacion_anual_cbt**, seguida por **variacion_anual_salarios_pct** y **log_densidad**. Estas variables aportan capacidad explicativa adicional al modelo, aunque con una influencia considerablemente menor respecto de la persistencia temporal del fenómeno.

Por último, **variacion_acceso_internet_pct** presenta una contribución reducida y **variacion_empleo_const_pct** muestra un impacto prácticamente nulo dentro del modelo final.

#### Dirección del efecto de las variables

El gráfico de dispersión SHAP permite analizar no solo la importancia de cada variable, sino también la dirección de su influencia sobre las predicciones.

En el caso de **tasa_lag1**, se observa una relación clara y consistente: valores elevados de la tasa histórica generan contribuciones SHAP positivas, aumentando la predicción futura de delitos contra la propiedad. Por el contrario, valores bajos de la tasa histórica producen contribuciones negativas o cercanas a cero. Este comportamiento evidencia una fuerte persistencia temporal del fenómeno.

Las restantes variables presentan efectos más moderados y concentrados cerca del valor cero, lo que indica que su influencia individual sobre las predicciones es significativamente menor. Aunque variables como la variación de la Canasta Básica Total, los salarios y la densidad poblacional aportan información al modelo, su capacidad explicativa resulta secundaria frente al peso de la trayectoria histórica del delito.

En conjunto, los resultados SHAP refuerzan la conclusión obtenida previamente mediante la importancia de variables y las pruebas de robustez: los delitos contra la propiedad presentan una marcada inercia temporal, complementada por factores estructurales territoriales y socioeconómicos que contribuyen a explicar parte de la variabilidad observada entre departamentos y años.


## 4.10 Validación temporal y robustez

La evaluación tradicional sobre un único conjunto de prueba permite estimar el desempeño general del modelo, pero no garantiza que dicho rendimiento se mantenga estable a lo largo del tiempo.

Por este motivo se implementó una validación temporal mediante un esquema rolling/expanding window, donde los modelos se entrenan utilizando únicamente información disponible hasta un determinado año y luego se evalúan sobre períodos posteriores.

Adicionalmente, se realizaron pruebas de robustez eliminando la variable `tasa_lag1`, con el objetivo de cuantificar cuánto depende la capacidad predictiva de la persistencia histórica del fenómeno.

*   **Robustez Confirmada:** La validación temporal confirma que ambos modelos son efectivos y robustos, mejorando su capacidad predictiva a medida que incorporan más datos. Random Forest demostró ser especialmente estable en los años posteriores.
