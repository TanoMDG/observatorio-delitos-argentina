```python
delitos_propiedad_lista = [
    "Hurtos",
    "Tentativas de hurto",
    "Robos (excluye los agravados por el resultado de lesiones y/o muertes)",
    "Robos agravados por el resultado de lesiones y/o muertes",
    "Tentativas de robo (excluye las agravadas por el res. de lesiones y/o muerte)",
    "Tentativas de robo agravado por el resultado de lesiones y/o muertes",
    "Daños (no incluye informáticos)",
    "Otros delitos contra la propiedad",
    "Estafas y defraudaciones (no incluye virtuales) y usura",
    "Estafas y defraudaciones asistidas virtualmente",
    "Extorsiones"
]
```

```python
df_propiedad = df[
    df["codigo_delito_snic_nombre"].isin(delitos_propiedad_lista)
]
```

```python
df_propiedad["codigo_delito_snic_nombre"].unique()
```

```python
df_agg = df_propiedad.groupby(
    [
        "provincia_id",
        "provincia_nombre",
        "departamento_id",
        "departamento_nombre",
        "anio"
    ]
).agg({
    "cantidad_hechos": "sum",
    "cantidad_victimas": "sum"
}).reset_index()
```

```python
df_agg = df_agg.rename(columns={
    "cantidad_hechos": "delitos_propiedad_hechos",
    "cantidad_victimas": "delitos_propiedad_victimas"
})
```

```python
import plotly.express as px

df_year = df_agg.groupby("anio", as_index=False)["delitos_propiedad_hechos"].sum()
fig = px.line(
    df_year,
    x="anio",
    y="delitos_propiedad_hechos",
    title="Evolución de delitos contra la propiedad",
    labels={"anio": "Año", "delitos_propiedad_hechos": "Hechos"},
    template="plotly_dark"
)
fig.show()
```
