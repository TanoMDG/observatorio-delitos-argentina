## Procesar variable IPC

```python


df_ipc = pd.read_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/raw/INDEC/serie_ipc_divisiones.csv",
                       encoding="latin1",
    sep=";")
```

```python
print(df_ipc.columns.tolist())
print(df_ipc.shape)
df_ipc.head(10)

```

```python
# Filtrar solo el nivel general
df_ipc = df_ipc[df_ipc["Descripcion"] == "NIVEL GENERAL"]
```

```python
# Reemplazar comas por puntos y convertir a float
df_ipc["Indice_IPC"] = df_ipc["Indice_IPC"].str.replace(",", ".").astype(float)
```

```python
# Convertir la columna "Periodo" a formato datetime
df_ipc["Periodo"] = df_ipc["Periodo"].astype(str)

df_ipc["fecha"] = pd.to_datetime(df_ipc["Periodo"], format="%Y%m")
df_ipc["anio"] = df_ipc["fecha"].dt.year
```

```python
# Crear un diccionario de mapeo para las regiones
map_region = {
    "GBA": "gba",
    "Pampeana": "pampeana",
    "Noreste": "nea",
    "Noroeste": "noa",
    "Cuyo": "cuyo",
    "Patagonia": "patagonia",
    "Nacional": "nacional"
}

df_ipc["region"] = df_ipc["Region"].map(map_region)
```

```python
# Agrupar por región y año, y calcular el promedio del índice IPC
df_ipc_anual = (
    df_ipc
    .groupby(["region", "anio"])["Indice_IPC"]
    .mean()
    .reset_index()
)
```

```python
# Filtrar para eliminar la región "nacional"
df_ipc_anual = df_ipc_anual [df_ipc_anual["region"] != "nacional"]
```

```python
# Guardar el dataset procesado
df_ipc_anual.to_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/processed/INDEC/ipc.csv", index=False)
```

## Procesar variable Salarios

```python
# Cargar el dataset de variación del índice de salarios
df_salarios = pd.read_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/raw/INDEC/variacion_indice_salarios.csv",
                          encoding="latin1", sep=";")
```

```python
print(df_salarios.shape)
print(df_salarios.columns.tolist())
df_salarios
```

```python
# Quedarme solo con las columnas necesarias.
df_salarios = df_salarios[["periodo", "v_i_a_indice_total"]].copy()

# Renombrar las columnas para que sean más descriptivas.

df_salarios.rename(columns={"periodo": "fecha", "v_i_a_indice_total": "variacion_anual_salarios_pct"}, inplace=True)

df_salarios.head(35)
```

```python
# Pasar fecha a formato datetime
df_salarios["fecha"] = pd.to_datetime(df_salarios["fecha"], dayfirst=True)
```

```python
# Convertir la variación anual de salarios a formato numérico

df_salarios["variacion_anual_salarios_pct"] = df_salarios["variacion_anual_salarios_pct"].astype(str).str.replace(",", ".", regex=False).astype(float)

df_salarios["variacion_anual_salarios_pct"] = pd.to_numeric(df_salarios["variacion_anual_salarios_pct"], errors='coerce').round(2)
```

```python
# Tomar año y mes de la fecha
df_salarios["anio"] = df_salarios["fecha"].dt.year
df_salarios["mes"] = df_salarios["fecha"].dt.month
```

```python
df_salarios.head(10)
```

```python
# tomar mes de Diciembre de cada año para obtener la variacion anual de salarios
df_salarios_anual = df_salarios[df_salarios["mes"] == 12][["anio", "variacion_anual_salarios_pct"]].copy()
```

```python
# Ordenar por año
df_salarios_anual = df_salarios_anual.sort_values("anio").reset_index(drop=True)
df_salarios_anual
```

La variación salarial anual se calcula utilizando la variación interanual del índice total correspondiente al mes de diciembre de cada año. Para 2016 no se dispone de valor interanual completo, por lo que se considera como dato faltante.

```python
# Guardar el dataframe procesado
df_salarios_anual.to_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/processed/INDEC/salarios_variacion_anual.csv", index=False)
```

# Procesar variable Canasta Básica Total

```python
# Cargar el dataset
df_cbt = pd.read_excel(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/raw/INDEC/Cba_y_Cbt.xlsx")

print(df_cbt.columns.tolist())
print(df_cbt.info())
df_cbt.head(19)
```

```python
# 1. Limpiar filas basura
df_cbt = df_cbt[pd.to_datetime(df_cbt["Mes"], errors="coerce").notna()].copy()
```

```python
# Renombrar columnas para que sean más descriptivas
df_cbt =df_cbt.rename(columns={"Mes": "fecha", "Canasta básica total": "cbt"})
```

```python
# Convertir a tipos fecha y numérico
df_cbt["fecha"] = pd.to_datetime(df_cbt["fecha"])

df_cbt["cbt"] = pd.to_numeric(df_cbt["cbt"], errors="coerce")
```

```python
# Crear año y mes
df_cbt["anio"] = df_cbt["fecha"].dt.year
df_cbt["mes"] = df_cbt["fecha"].dt.month

# calculo la variacion

df_cbt = df_cbt.sort_values("fecha")
df_cbt["variacion_anual_cbt"] = (df_cbt["cbt"].pct_change(periods=12) * 100).round(2)

df_cbt.tail(20)
```

```python
# Llevar a variacion anual de la cbt
df_cbt_anual = df_cbt[df_cbt["mes"]==12][["anio", "variacion_anual_cbt"]].copy()

df_cbt_anual = df_cbt_anual.sort_values("anio").reset_index(drop=True)
df_cbt_anual
```

```python
# Guardar el dataframe procesado

df_cbt_anual.to_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/processed/INDEC/cbt_variacion_anual.csv")
```

# Procesar variable Puestos de Trabajo en la construcción

```python
# cargar dataset
df_trab_constr = pd.read_excel(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/raw/INDEC/Puestos_trab_const.xlsx")

print(df_trab_constr.columns.tolist())
print(df_trab_constr.info())
df_trab_constr.head(30)
```

```python
# Eliminar filas basura
df_trab_constr = df_trab_constr.dropna(subset=["Puestos de trabajo"]).copy()
print(df_trab_constr.columns.tolist())
print(df_trab_constr.info())
df_trab_constr.head(30)
```

```python
# renombrar columna
df_trab_constr = df_trab_constr.rename(columns={"Unnamed: 1": "mes", "Variación porcentual": "Variacion_pct_mes", "Unnamed: 4": "Variacion_pct_anual"})
```

```python
# Reconstruir columna del año
df_trab_constr["anio"] = df_trab_constr["Período"].str.extract(r"(\d{4})")
df_trab_constr["anio"] = df_trab_constr["anio"].ffill()
df_trab_constr["anio"] = pd.to_numeric(df_trab_constr["anio"], errors="coerce")
df_trab_constr.head(30)
```

```python
# mapear meses a números
map_meses = {
    "Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4,
    "Mayo": 5, "Junio": 6, "Julio": 7, "Agosto": 8,
    "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12
}

df_trab_constr["mes_num"] = df_trab_constr["mes"].map(map_meses)
df_trab_constr.head()
```

```python
# Limpiar variacion interanual de puestos de trabajo
df_trab_constr["variacion_pct"] = pd.to_numeric(df_trab_constr["Variacion_pct_anual"].astype(str).str.replace(",",".", regex=False).replace("///", None))
df_trab_constr["variacion_pct"] = pd.to_numeric(df_trab_constr["variacion_pct"], errors="coerce")
df_trab_constr
```

```python
# Quedarme solo con diciembre de cada año
df_trab_constr_anual = df_trab_constr[df_trab_constr["mes_num"]==12][["anio", "variacion_pct"]].copy()
```

```python
# Ordenar
df_trab_constr_anual = df_trab_constr_anual.sort_values("anio").reset_index(drop=True)
df_trab_constr_anual
```

```python
# Renombrar columna PRINCIPAL
df_trab_constr_anual = df_trab_constr_anual.rename(columns={"variacion_pct": "variacion_empleo_const_pct"})
```

```python
# guardar el dataframe procesado
df_trab_constr_anual.to_csv(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/processed/INDEC/empleo_const_anual.csv")
```

# Procesar variable de Accesos a Internet


```python
df_acceso_internet = pd.read_excel(r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/raw/INDEC/Accesos_internet_prov.xlsx")
print(df_acceso_internet.columns.tolist())
print(df_acceso_internet.info())
df_acceso_internet.head(20)
```

```python
# Reconstruir correctamente los nombres de columnas del Excel
# Tomo las dos primeras filas como encabezados
fila_anios = df_acceso_internet.iloc[0]
fila_meses = df_acceso_internet.iloc[1]

# Construir nombres de columnas nuevos
nuevas_columnas = []

for i in range(len(df_acceso_internet.columns)):
    # La primera columna es la provincia/jurisdicción
    if i == 0:
        nuevas_columnas.append("jurisdiccion")
    else:
        anio = fila_anios.iloc[i]
        mes = fila_meses.iloc[i]
        nuevas_columnas.append(f"{anio}_{mes}")

# Aplicamos los nuevos nombres
df_acceso_internet.columns = nuevas_columnas

# Miramos cómo quedó
print(df_acceso_internet.columns.tolist()[:15])
df_acceso_internet.head(5)
```

```python
df_acceso_internet = df_acceso_internet.iloc[3:].reset_index(drop=True)
df_acceso_internet.head(10)
```

```python
# Pasar de formato ancho a formato largo
df_acceso_internet_long = df_acceso_internet.melt(
    id_vars="jurisdiccion",
    var_name="periodo",
    value_name="accesos_internet"
)

print(df_acceso_internet_long.shape)
df_acceso_internet_long.head(10)
```

```python
# Separar período en año y mes
df_acceso_internet_long[["anio","mes_texto"]] = df_acceso_internet_long["periodo"].str.split("_", expand=True)

# Convertir año a número
df_acceso_internet_long["anio"] = pd.to_numeric(df_acceso_internet_long["anio"])

df_acceso_internet_long.head(10)
```

```python
# Mapear meses a números
map_meses = {
    "Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4,
    "Mayo": 5, "Junio": 6, "Julio": 7, "Agosto": 8,
    "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12
}

df_acceso_internet_long["mes"] = df_acceso_internet_long["mes_texto"].map(map_meses)

df_acceso_internet_long.head(10)
```

```python
# Aseguro que accesos_internet sea numerico
df_acceso_internet_long["accesos_internet"] = pd.to_numeric(df_acceso_internet_long["accesos_internet"], errors="coerce")
```

```python
# Ordenar
df_acceso_internet_long = df_acceso_internet_long.sort_values(["jurisdiccion", "anio", "mes"]).reset_index(drop=True)
```

```python
# Calcular variacion anual por provincia
df_acceso_internet_long["variacion_acceso_internet_pct"]= df_acceso_internet_long.groupby("jurisdiccion")["accesos_internet"].pct_change(periods=12)*100
```

```python
# Pasar a anual quedandome con diciembre de cada año
df_acceso_internet_anual = df_acceso_internet_long[
    df_acceso_internet_long["mes"] == 12
][
    ["jurisdiccion", "anio", "variacion_acceso_internet_pct"]
].copy()

df_acceso_internet_anual = df_acceso_internet_anual.sort_values(
    ["jurisdiccion", "anio"]
).reset_index(drop=True)
df_acceso_internet_anual
```

```python
# Guardar el dataframe procesado
df_acceso_internet_anual.to_csv(
    r"/content/drive/MyDrive/Practica_Profesionalizante_3/Copia_pre_proyecto/data/processed/INDEC/variacion_accesos_internet_provincias_anual.csv",
    index=False
)
```
