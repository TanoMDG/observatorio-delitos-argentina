## 🔗 Análisis de la relación entre variables socioeconómicas y delitos contra la propiedad

### 1. Inflación (IPC) y delito

El gráfico muestra una **relación positiva débil** entre el nivel del índice de precios (IPC) y la tasa de delitos.

Observaciones:
- La línea de tendencia tiene pendiente levemente ascendente.
- Existe una **alta dispersión de los datos**, con muchos valores para un mismo nivel de IPC.
- Se observan agrupamientos por año (bandas verticales), lo que indica que el IPC varía más entre años que entre departamentos.
- A mayor inflación, podría existir una **leve tendencia al aumento del delito**, aunque no es una relación fuerte.
- La inflación por sí sola **no explica significativamente la variabilidad del delito a nivel territorial**.

---

### 2. Salarios y delito

El gráfico presenta una **relación positiva muy débil** entre la variación salarial y la tasa de delitos.

Observaciones:
- La pendiente de la línea de tendencia es levemente positiva.
- La dispersión es alta, similar al caso del IPC.
- Se observan valores extremos de delito en distintos niveles salariales.
- No se evidencia una relación clara entre mejoras salariales y reducción del delito.
- Esto sugiere que el delito **no responde directamente a la evolución de los salarios nominales**, o que el efecto está mediado por otras variables.

---

### 3. Empleo en construcción y delito

Este gráfico muestra una **relación prácticamente nula** entre la variación del empleo en construcción y la tasa de delitos.

Observaciones:
- La línea de tendencia es casi horizontal.
- La nube de puntos no presenta patrón claro.
- Existen valores altos de delito tanto en contextos de crecimiento como de caída del empleo.
- El empleo en construcción, al menos en esta forma agregada, **no parece tener un impacto directo sobre el delito**.
- Es posible que esta variable no capture adecuadamente las dinámicas laborales relevantes para el fenómeno.

---

### 4. Costo de vida (CBT) y delito

El gráfico evidencia una **relación positiva débil a moderada** entre el aumento del costo de vida (CBT) y la tasa de delitos.

Observaciones:
- La pendiente de la línea de tendencia es más marcada que en los otros casos.
- Se mantiene una alta dispersión, pero con una tendencia más visible.
- Los valores más altos de CBT coinciden con niveles más elevados de delito.
- El aumento del costo de vida podría estar **más asociado al incremento del delito** que otras variables analizadas.
- Esto sugiere que variables vinculadas a la **presión económica sobre los hogares** pueden tener mayor relevancia explicativa.

---

## 🧠 Conclusión general del análisis de relaciones

- Las relaciones entre variables macroeconómicas y delito son en general **débiles y con alta dispersión**.
- No se observan vínculos lineales fuertes, lo que indica que el fenómeno es **multicausal y complejo**.
